```python
## Primary key validation

primary_keys = {
    "customers": "customer_id",
    "orders": "order_id",
    "customer_features": "customer_id",
    "customer_unit_economics": "customer_id",
    "subscription_history": "subscription_id"
}

for table, key in primary_keys.items():

    df = datasets[table]

    total_rows = len(df)
    unique_values = df[key].nunique()
    duplicate_count = df[key].duplicated().sum()
    null_count = df[key].isna().sum()

    print(f"\nTable: {table}")
    print(f"Primary Key: {key}")
    print(f"Total Rows: {total_rows}")
    print(f"Unique Values: {unique_values}")
    print(f"Duplicate Values: {duplicate_count}")
    print(f"Null Values: {null_count}")
```

```python
## Foreign key validation

foreign_keys = {
    "orders": "customer_id",
    "customer_features": "customer_id",
    "customer_unit_economics": "customer_id",
    "subscription_history": "customer_id"
}

customer_ids = set(customers["customer_id"])

for table, key in foreign_keys.items():

    df = datasets[table]

    unmatched = df.loc[~df[key].isin(customer_ids), key]

    print(f"\nTable: {table}")
    print(f"Total Records: {len(df)}")
    print(f"Unmatched Customer IDs: {len(unmatched)}")
    print(f"Unique Unmatched IDs: {unmatched.nunique()}")
```

```python
## Missing value analysis

for name, df in datasets.items():

    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df)) * 100

    missing_summary = pd.DataFrame({
        "Missing Values": missing_count,
        "Missing Percentage": missing_percentage.round(2)
    })

    missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

    print(f"\n{'='*60}")
    print(f"TABLE: {name}")
    print(f"{'='*60}")

    if missing_summary.empty:
        print("No missing values found.")
    else:
        display(missing_summary)
```

```python
## Validate net_revenue
## net_revenue = revenue - discount - refund

# Recalculate net revenue
orders["calculated_net_revenue"] = (
    orders["revenue"]
    - orders["discount"]
    - orders["refund_amount"]
)

# Calculate difference from recorded net revenue
orders["net_revenue_difference"] = (
    orders["net_revenue"]
    - orders["calculated_net_revenue"]
)

# Identify mismatches, allowing for floating-point precision
net_revenue_mismatches = orders[
    ~np.isclose(
        orders["net_revenue"],
        orders["calculated_net_revenue"],
        atol=0.01
    )
]

print("Total orders:", len(orders))
print("Net revenue mismatches:", len(net_revenue_mismatches))

print("\nDifference statistics:")
display(orders["net_revenue_difference"].describe())
```


### Net Revenue Validation

Formula:
Net Revenue = Revenue - Discount - Refund Amount

- Total orders checked: 1,126,821
- Mismatches: 0
- Floating-point differences were negligible.

**Result:** Passed. The recorded net_revenue values are consistent with the independently calculated values.


```python
## Gross margin validation
## gross margin = net_revenue - cogs - shipping cost

orders["calculated_gross_margin"] = (
    orders["net_revenue"]
    - orders["cost_of_goods"]
    - orders["shipping_cost"]
)

orders["gross_margin_difference"] = (
    orders["gross_margin"]
    - orders["calculated_gross_margin"]
)

gross_margin_mismatches = orders[
    ~np.isclose(
        orders["gross_margin"],
        orders["calculated_gross_margin"],
        atol=0.01
    )
]

print("Total orders:", len(orders))
print("Gross margin mismatches:", len(gross_margin_mismatches))

print("\nDifference statistics:")
display(orders["gross_margin_difference"].describe())
```


### Gross Margin Validation

Formula:
Gross Margin = Net Revenue - COGS - Shipping Cost

- Total orders checked: 1,126,821
- Mismatches: 0
- Floating-point differences were negligible.

**Result:** Passed.

**Business observation:** The supplied gross_margin excludes delivery_region_cost and payment_processing_fee. These costs must be considered separately when calculating adjusted contribution.


```python
## Refund and order status validation

refund_status_summary = (
    orders.groupby("order_status")
    .agg(
        total_orders=("order_id", "count"),
        orders_with_refunds=(
            "refund_amount",
            lambda x: (x > 0).sum()
        ),
        total_refunds=("refund_amount", "sum"),
        average_refund=("refund_amount", "mean")
    )
)

display(refund_status_summary)
```

```python
## Are refunds financially reasonable?
## refundable amount = revenue - discount

# Calculate the order value after discounts
orders["value_after_discount"] = (
    orders["revenue"] - orders["discount"]
)

# Identify refunds exceeding the post-discount value
excessive_refunds = orders[
    orders["refund_amount"] > orders["value_after_discount"]
]

# Identify orders with negative net revenue
negative_net_revenue = orders[
    orders["net_revenue"] < 0
]

# Identify returned orders without refunds
returned_without_refunds = orders[
    (orders["order_status"] == "Returned") &
    (orders["refund_amount"] == 0)
]

print("Refunds exceeding post-discount order value:",
      len(excessive_refunds))

print("Orders with negative net revenue:",
      len(negative_net_revenue))

print("Returned orders without refunds:",
      len(returned_without_refunds))
```

```python
## Investigate the magnitude of the discrepancy

# Calculate the excess refund amount
excessive_refunds = excessive_refunds.copy()

excessive_refunds["excess_refund"] = (
    excessive_refunds["refund_amount"]
    - excessive_refunds["value_after_discount"]
)

print("Excess refund statistics:")
display(excessive_refunds["excess_refund"].describe())

print("\nSample excessive refund records:")

display(
    excessive_refunds[
        [
            "order_id",
            "order_status",
            "revenue",
            "discount",
            "refund_amount",
            "value_after_discount",
            "excess_refund"
        ]
    ].head(10)
)
```

```python
# Recheck excessive refunds with a tolerance of 0.01

excessive_refunds = orders[
    orders["refund_amount"] >
    (orders["value_after_discount"] + 0.01)
]

print("Actual excessive refunds:", len(excessive_refunds))
```

- Initial excessive refund check identified 15,968 apparent discrepancies.
- Investigation confirmed these were floating-point precision artefacts.
- After applying a ₹0.01 tolerance, actual excessive refunds: 0.
- Orders with negative net revenue: 0.

**Result:** Passed.
